1. Dada una variable edad, imprime "Mayor de edad" o "Menor de edad" usando un if ternario.

```python
edad = int(input("Ingrese su edad: "))
print("Usted puede ir preso" if edad >= 18 else "Usted es menor de edad, no puede ir preso")
```

2. Escribe un programa que reciba una nota numérica (0-100) y clasifique el resultado: 90-100 es "A", 80-89 es "B", 70-79 es "C", y menos de 70 es "F"

```python
Puntaje = float(input("Ingrese el puntaje obtenido en su evaluación: "))
if Puntaje >= 90:
    print("Clasificación: A")
elif Puntaje >= 80:
    print("Clasificación: B")
elif Puntaje >= 70:
    print("Clasificación: C")
else:
    print("Clasificación: F")
```

3. Tienes una variable temperatura. Si es mayor a 30, imprime "Hace calor", si está entre 15 y 30, imprime "Clima agradable", y si es menor, "Hace frío"

```python
temperatura = float(input("Ingrese la temperatura en grados Celsius: "))
if temperatura > 30:
    print("Hace calor")
elif temperatura >= 15:
    print("Clia agradable")
else:
    print("Hace frío")
```

4. Dado un número n, usa un if ternario para asignar a una variable la cadena "Par" o "Impar".

```python
n = int(input("Ingrese un número: "))
resultado = "Par" if n % 2 == 0 else "Impar"
print(resultado)
```
